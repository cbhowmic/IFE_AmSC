# Batch RHINO-to-BP5 conversion

This notebook calls the maintained `convert_scenarios` implementation to convert RHINO pickle outputs into one openPMD/ADIOS2 BP5 product per run. Run it in an environment where the `rhino-amsc` package and its `io` dependencies are installed, then update the paths and scenarios below before starting the conversion.

```python
from pathlib import Path

from rhino.shim.rhinoWrite_multiple import convert_scenarios
```

```python
ROOT_PATH = Path(
    "/global/cfs/cdirs/m3239/2026_FES-AmSC/data/rhino/Surrogate Data"
)
OUTPUT_ROOT = Path(
    "/global/cfs/cdirs/m3239/2026_FES-AmSC/data/rhino/surrogate_bp_output"
)
SCENARIOS = ["2026-04-29", "2026-04-30", "2026-05-01"]

# Add (scenario, HH-MM-SS prefix) pairs only when a run must be omitted.
SKIP_RUNS = set()
```

```python
convert_scenarios(
    root_path=ROOT_PATH,
    scenarios=SCENARIOS,
    output_root=OUTPUT_ROOT,
    skip_runs=SKIP_RUNS,
)
```
